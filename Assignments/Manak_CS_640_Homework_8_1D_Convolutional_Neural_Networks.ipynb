## Q1

Consider a 1D convolution of a vector $[6, 3, 6]$ and filter $[2, 2, 1]$ with stride $1$. Additionally, a ReLU activation is applied after the convolution. Calculate the final result.

**Answer**

Length of input vector => $L=3$.

Length of the filter is $F_L=3$.

The stride is $S=1$.

$L_{out} = \frac{L - F_L}{S} + 1 = \frac{3 - 3}{1} + 1 = 1$

There will be one output value.

the first window $[6, 3, 6]$:
   
Convolution result
   
$= (6 \times 2) + (3 \times 2) + (6 \times 1)$
   
$= 12 + 6 + 6 = 24$

ReLU activation ($\text{ReLU}(x) = \max(0, x)$):
$\text{ReLU}(24) = 24$

the final result is $[24]$.


## Q2

Consider a 1D convolution of a vector $[6, 3, 6]$ and filter $[2, 2, 1]$ with stride $1$ and padding size $1$. Additionally, a ReLU activation is applied after the convolution. Calculate the final result.

Hint: what is the key difference between this question and Q1?

**Answer**

padding means our input vector becomes $[0, 6, 3, 6, 0]$.

Now, with this padded vector (length 5) and filter (length 3), and a stride of 1, our output length will be $(5 - 3)/1 + 1 = 3$ values.

 So, we'll get three outputs:

1.  **First window** $[0, 6, 3]$:
    (0 * 2) + (6 * 2) + (3 * 1) = 0 + 12 + 3 = 15

    Applying ReLU : ReLU(15) = 15

2.  **Second window** $[6, 3, 6]$:
    (6 * 2) + (3 * 2) + (6 * 1) = 12 + 6 + 6 = 24

    Applying ReLU: ReLU(24) = 24

3.  **Third window** $[3, 6, 0]$:
    (3 * 2) + (6 * 2) + (0 * 1) = 6 + 12 + 0 = 18
    
    Applying ReLU: ReLU(18) = 18

final result is $[15, 24, 18]$.

The key difference from Q1 is that padding adds zeros around the input, making the output longer and letting the filter 'see' the edges of the original vector better

## Q3

Consider a 1D convolution of a vector of size $16$ and a filter of size $3$ with stride $2$. Calculate the size of the convolution output with padding size $1$ (per side).

**Answer**

$L_{out} = \lfloor \frac{\text{Input Size} + 2 \times \text{Padding} - \text{Filter Size}}{\text{Stride}} \rfloor + 1$


$L_{out} = \lfloor \frac{16 + 2 \times 1 - 3}{2} \rfloor + 1$

$L_{out} = \lfloor \frac{15}{2} \rfloor + 1$

$L_{out} = 7 + 1$

$L_{out} = 8$

So, the final output size will be 8.

## Q4

Consider a 1D convolution layer with input dimension $1280$, $12$ filters of size $3$ and stride $3$, and no padding. Calculate the number of parameters. Assume that there is no bias.

**Answer**

$Number of Parameters = (Filter Size × Input Channels) × Number of Filters$


$Parameters = (3 × 1280) × 12$

$Parameters = 3840 × 12$

$Parameters = 46080$

total number of parameters for this layer is $46080$
