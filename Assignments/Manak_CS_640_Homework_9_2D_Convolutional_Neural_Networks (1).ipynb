## Question 5

Let $L$ be a pooling layer in a CNN, and let $x$ and $y$ be the input and output of $L$ respectively. Is it possible that size($x$) < size($y$)? Briefly explain.

**Answer**

Pooling layers can't increase the spatial or output size, either keep it the same or  can reduce it, since their main job is downsampling.

Each K×K pooling window gives just one output value, so the result is always fewer or the same number of pixels as the input, never more.
